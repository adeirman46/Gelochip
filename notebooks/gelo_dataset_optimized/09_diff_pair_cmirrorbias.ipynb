# 09 — Differential Pair with Current Mirror Bias

**Dataset:** `datasets/diff_pair_cmirrorbias/`

Topology: 2 NMOS current mirror (tail) + 4 NMOS diff pair (ABBA)
- CM tail: M_ref(G=D=IBIAS) + M_copy(G=IBIAS,D=VTAIL)
- Diff pair (×4 ABBA): G=VP/VN, D=VDD1/VDD2, S=VTAIL

Nodes: `VP VN VDD1 VDD2 IBIAS VSS`

```python
import sys, os
os.environ.setdefault('PDK_ROOT', os.path.expanduser('~/pdks'))
sys.path.insert(0, os.path.abspath('../../src/gelochip'))
import gelochip.gl as gl
gl.reload()  # pick up latest code without restarting kernel
```

```python
vp    = gl.Net('vp')
vn    = gl.Net('vn')
vdd1  = gl.Net('vdd1')
vdd2  = gl.Net('vdd2')
ibias = gl.Net('ibias')   # external bias current → reference diode
vtail = gl.Net('vtail')   # mirror copy output = diff pair tail

# Current mirror tail source
m_ref  = gl.nmos(w=4.0, fingers=1, g=ibias, d=ibias, s=gl.gnd)  # diode ref
m_copy = gl.nmos(w=4.0, fingers=1, g=ibias, d=vtail, s=gl.gnd)  # mirror copy → tail

# 4-NMOS diff pair ABBA
m_tl = gl.nmos(w=3.0, fingers=2, g=vp, d=vdd1, s=vtail)  # A top
m_bl = gl.nmos(w=3.0, fingers=2, g=vp, d=vdd1, s=vtail)  # A bot
m_tr = gl.nmos(w=3.0, fingers=2, g=vn, d=vdd2, s=vtail)  # B top
m_br = gl.nmos(w=3.0, fingers=2, g=vn, d=vdd2, s=vtail)  # B bot

chip = gl.build(m_ref, m_copy, m_tl, m_bl, m_tr, m_br, name='diff_pair_cmirrorbias')
chip.show()
chip.drc()
chip.sim()
```
