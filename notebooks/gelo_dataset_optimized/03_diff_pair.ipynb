# 03 — Differential Pair

Built from scratch: two matched NMOS transistors sharing a tail node.

```
     vip ──G(M+)D── vop
               S
              vtail
               S
     vim ──G(M-)D── vom
```

## Basic differential pair (no tail source)

````python
vip  = gl.Net('vip')
vim  = gl.Net('vim')
vtail = gl.Net('vtail')
vop  = gl.Net('vop')
vom  = gl.Net('vom')

mn_p = gl.nmos(w=3.0, fingers=2, g=vip, d=vop, s=vtail)
mn_m = gl.nmos(w=3.0, fingers=2, g=vim, d=vom, s=vtail)

chip = gl.build(mn_p, mn_m, name='diff_pair')
chip.show()
chip.drc()
chip.sim()
````

## With tail current source (gate-biased NMOS)

````python
vbias = gl.Net('vbias')
vip2  = gl.Net('vip2')
vim2  = gl.Net('vim2')
vtail2 = gl.Net('vtail2')
vop2  = gl.Net('vop2')
vom2  = gl.Net('vom2')

mn_tail = gl.nmos(w=4.0, fingers=2, g=vbias, d=vtail2, s=gl.gnd)
mn_p2   = gl.nmos(w=3.0, fingers=2, g=vip2,  d=vop2,   s=vtail2)
mn_m2   = gl.nmos(w=3.0, fingers=2, g=vim2,  d=vom2,   s=vtail2)

chip2 = gl.build(mn_tail, mn_p2, mn_m2, name='diff_pair_with_tail')
chip2.show()
chip2.drc()
chip2.sim()
````

## Module style — parametric diff pair

````python
class DiffPair(gl.Module):
    def __init__(self, wn=3.0, wt=4.0, fn=2):
        super().__init__(gl.gf180)
        self.wn, self.wt, self.fn = wn, wt, fn

    def layout(self):
        vip   = self.pin('vip')
        vim   = self.pin('vim')
        vbias = self.pin('vbias')
        vop   = self.pin('vop')
        vom   = self.pin('vom')
        vtail = gl.Net('vtail')
        gl.nmos(w=self.wt, fingers=2,       g=vbias, d=vtail, s=gl.gnd)
        gl.nmos(w=self.wn, fingers=self.fn, g=vip,   d=vop,   s=vtail)
        gl.nmos(w=self.wn, fingers=self.fn, g=vim,   d=vom,   s=vtail)

dp = DiffPair(wn=3, wt=4, fn=2).build()
dp.show()
dp.drc()
````
