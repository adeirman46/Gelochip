# 15 — Row CS Amplifier (Diff-to-Single + Common-Source)

**Dataset:** `datasets/row_csamplifier_diff_to_single_ended_converter/`

Topology: diff-to-single (4 PMOS) + PMOS CS amplifier (2 PMOS)
- diff-to-single takes VIN1 → produces VIN2 (internal)
- CS amp: G=VIN2, D=VOUT, S=VDD (left+right symmetric pair)

Nodes: `VIN1 VIN2 VOUT VSS VSS2`

```python
import sys, os
os.environ.setdefault('PDK_ROOT', os.path.expanduser('~/pdks'))
sys.path.insert(0, os.path.abspath('../../src/gelochip'))
import gelochip.gl as gl
gl.reload()  # pick up latest code without restarting kernel
```

```python
vin1 = gl.Net('vin1')  # diff-to-single input
vin2 = gl.Net('vin2')  # diff-to-single output → CS amp gate
vout = gl.Net('vout')  # CS amp drain output
v1   = gl.Net('v1')    # diff-to-single internal
vss2 = gl.Net('vss2')  # diff-to-single internal

# ── diff-to-single converter (same as notebook 14, VIN=VIN1, VOUT=VIN2) ─
mp_t1 = gl.pmos(w=6.0, fingers=6, g=vin1, d=v1,   s=gl.vdd)
mp_t2 = gl.pmos(w=6.0, fingers=6, g=vin1, d=vss2, s=gl.vdd)
mp_b1 = gl.pmos(w=6.0, fingers=6, g=vin1, d=vin1, s=v1)
mp_b2 = gl.pmos(w=6.0, fingers=6, g=vin1, d=vin2, s=vss2)   # → VIN2

# ── PMOS CS amplifier: G=VIN2, D=VOUT, S=VDD (left + right symmetric) ─
mp_cs_l = gl.pmos(w=7.0, fingers=10, g=vin2, d=vout, s=gl.vdd)
mp_cs_r = gl.pmos(w=7.0, fingers=10, g=vin2, d=vout, s=gl.vdd)

chip = gl.build(mp_t1, mp_t2, mp_b1, mp_b2, mp_cs_l, mp_cs_r,
                name='row_csamplifier')
chip.show()
chip.drc()
chip.sim()
```
