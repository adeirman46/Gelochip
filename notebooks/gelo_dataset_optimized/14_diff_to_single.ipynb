# 14 — Differential to Single-Ended Converter

**Dataset:** `datasets/differential_to_single_ended_converter/`

Topology: 4 PMOS cross-coupled load
```
VDD ─ XTOP1(D=V1, G=VIN)  ─ XTOP2(D=VSS2, G=VIN)
         │                         │
       V1 ─ XBOT1(D=VIN,G=VIN)    VSS2 ─ XBOT2(D=VOUT,G=VIN)
                     └───VIN (diff input)
```
Nodes: `VIN VOUT`  (V1, VSS2 are internal)

````python
import sys, os
os.environ.setdefault('PDK_ROOT', os.path.expanduser('~/pdks'))
sys.path.insert(0, os.path.abspath('../../src/gelochip'))
import gelochip.gl as gl
gl.reload()  # pick up latest code without restarting kernel
````

````python
vin  = gl.Net('vin')   # receives VDD1 from NMOS diff pair (left drain)
vout = gl.Net('vout')  # receives VDD2 from NMOS diff pair (right drain)
v1   = gl.Net('v1')    # internal node
vss2 = gl.Net('vss2')  # internal node

# 4 PMOS cross-coupled (SPICE order D G S B)
mp_t1 = gl.pmos(w=6.0, fingers=6, g=vin,  d=v1,   s=gl.vdd)  # XTOP1
mp_t2 = gl.pmos(w=6.0, fingers=6, g=vin,  d=vss2, s=gl.vdd)  # XTOP2
mp_b1 = gl.pmos(w=6.0, fingers=6, g=vin,  d=vin,  s=v1)      # XBOT1 diode (S=V1)
mp_b2 = gl.pmos(w=6.0, fingers=6, g=vin,  d=vout, s=vss2)    # XBOT2 mirror

chip = gl.build(mp_t1, mp_t2, mp_b1, mp_b2, name='diff_to_single')
chip.show()
chip.drc()
chip.sim()
````
