# 02 — Current Mirror

Built from scratch using `gl.nmos()` and `gl.pmos()` primitives.

```
Ref:  M_ref  (gate=drain=vbias, source=GND) ← diode-connected
Copy: M_copy (gate=vbias,       drain=iout, source=GND)
```

````python
import sys, os
os.environ.setdefault('PDK_ROOT', os.path.expanduser('~/pdks'))
sys.path.insert(0, os.path.abspath('../../src/gelochip'))
import gelochip.gl as gl
gl.reload()  # pick up latest code without restarting kernel
print('Imports OK')
````

## NMOS 1:1 current mirror

````python
vbias = gl.Net('vbias')
iout  = gl.Net('iout')

m_ref  = gl.nmos(w=4.0, fingers=1, g=vbias, d=vbias, s=gl.gnd) 
m_copy = gl.nmos(w=4.0, fingers=1, g=vbias, d=iout,  s=gl.gnd)

chip = gl.build(m_ref, m_copy, name='ncmirror_1to1')
chip.show()
chip.drc()
chip.sim()
````

## NMOS 1:2 ratio — double fingers on copy

````python
vbias2 = gl.Net('vbias2')
iout2  = gl.Net('iout2')

m_ref2  = gl.nmos(w=4.0, fingers=1, g=vbias2, d=vbias2, s=gl.gnd)
m_copy2 = gl.nmos(w=4.0, fingers=2, g=vbias2, d=iout2,  s=gl.gnd)  # 2x fingers → 1:2

chip2 = gl.build(m_ref2, m_copy2, name='ncmirror_1to2')
chip2.show()
chip2.drc()
````

## PMOS current source mirror

````python
vbias_p = gl.Net('vbias_p')
iout_p  = gl.Net('iout_p')

mp_ref  = gl.pmos(w=4.0, fingers=1, g=vbias_p, d=vbias_p, s=gl.vdd)  # diode-connected
mp_copy = gl.pmos(w=4.0, fingers=1, g=vbias_p, d=iout_p,  s=gl.vdd)

chip3 = gl.build(mp_ref, mp_copy, name='pcmirror_1to1')
chip3.show()
chip3.drc()
chip3.sim()
````

## Module style — parametric, reusable

````python
class CurrentMirror(gl.Module):
    def __init__(self, w=4.0, ratio=2, n_or_p='n'):
        super().__init__(gl.gf180)
        self.w, self.ratio, self.n_or_p = w, ratio, n_or_p

    def layout(self):
        vbias = self.pin('vbias')
        iout  = self.pin('iout')
        fet   = gl.nmos if self.n_or_p == 'n' else gl.pmos
        rail  = gl.gnd  if self.n_or_p == 'n' else gl.vdd
        fet(w=self.w, fingers=1,          g=vbias, d=vbias, s=rail)
        fet(w=self.w, fingers=self.ratio,  g=vbias, d=iout,  s=rail)

cm = CurrentMirror(w=4, ratio=4, n_or_p='n').build()
cm.show()
cm.drc()
````
