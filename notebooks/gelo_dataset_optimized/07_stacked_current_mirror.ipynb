# 07 — Stacked (Cascode) Current Mirror

**Dataset:** `datasets/stacked_current_mirror/`

Topology: 4 NMOS (2-stack cascode)
- Ref bottom: G=D=VBIAS, S=GND (diode)
- Ref top:    G=D=VCAS,  S=VBIAS (diode)
- Copy bottom: G=VBIAS, D=VMID, S=GND
- Copy top:    G=VCAS,  D=IOUT, S=VMID

Nodes: `VBIAS VCAS IOUT`

```python
import sys, os
os.environ.setdefault('PDK_ROOT', os.path.expanduser('~/pdks'))
sys.path.insert(0, os.path.abspath('../../src/gelochip'))
import gelochip.gl as gl
gl.reload()  # pick up latest code without restarting kernel
```

```python
vbias = gl.Net('vbias')   # bottom reference bias
vcas  = gl.Net('vcas')    # cascode bias
vmid  = gl.Net('vmid')    # copy stack mid-node
iout  = gl.Net('iout')

# Reference stack (both diode-connected)
m_rb = gl.nmos(w=4.0, fingers=1, g=vbias, d=vbias, s=gl.gnd)  # bottom diode
m_rt = gl.nmos(w=4.0, fingers=1, g=vcas,  d=vcas,  s=vbias)   # top diode

# Copy stack (mirrors the reference biases)
m_cb = gl.nmos(w=4.0, fingers=1, g=vbias, d=vmid,  s=gl.gnd)  # bottom mirror
m_ct = gl.nmos(w=4.0, fingers=1, g=vcas,  d=iout,  s=vmid)    # top mirror

chip = gl.build(m_rb, m_rt, m_cb, m_ct, name='stacked_current_mirror')
chip.show()
chip.drc()
chip.sim()
```
