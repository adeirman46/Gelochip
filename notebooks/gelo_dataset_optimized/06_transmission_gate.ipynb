# 06 — Transmission Gate

**Dataset:** `datasets/transmission_gate/`

Topology: NMOS + PMOS in parallel
- NMOS: G=VGN, D=VOUT, S=VIN (passes when VGN=HIGH)
- PMOS: G=VGP, D=VOUT, S=VIN (passes when VGP=LOW)

Nodes: `VIN VOUT VGN VGP`

```python
import sys, os
os.environ.setdefault('PDK_ROOT', os.path.expanduser('~/pdks'))
sys.path.insert(0, os.path.abspath('../../src/gelochip'))
import gelochip.gl as gl
gl.reload()  # pick up latest code without restarting kernel
```

```python
vin  = gl.Net('vin')
vout = gl.Net('vout')
vgn  = gl.Net('vgn')   # NMOS gate enable (HIGH=pass)
vgp  = gl.Net('vgp')   # PMOS gate enable_b (LOW=pass)

mn = gl.nmos(w=2.0, fingers=1, g=vgn, d=vout, s=vin)
mp = gl.pmos(w=2.0, fingers=1, g=vgp, d=vout, s=vin)

chip = gl.build(mn, mp, name='transmission_gate')
chip.show()
chip.drc()
chip.sim()
```
