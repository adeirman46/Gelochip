# 13 — P-Block (PMOS Super-Class AB Output Stage)

**Dataset:** `datasets/p_block/`

Topology: 4 PMOS interdigitized (2x2), all sharing gate MA_G
- MA_1, MA_2: G=MA_G, D=MA_1_D/MA_2_D, S=VDD
- MB_1, MB_2: G=MA_G, D=MB_1_D/MB_2_D, S=VDD

Nodes: `MA_G MA_1_D MA_2_D MB_1_D MB_2_D VDD`

```python
import sys, os
os.environ.setdefault('PDK_ROOT', os.path.expanduser('~/pdks'))
sys.path.insert(0, os.path.abspath('../../src/gelochip'))
import gelochip.gl as gl
gl.reload()  # pick up latest code without restarting kernel
```

```python
ma_g   = gl.Net('ma_g')    # shared gate (driven by N-block output)
ma_1d  = gl.Net('ma_1d')
ma_2d  = gl.Net('ma_2d')
mb_1d  = gl.Net('mb_1d')
mb_2d  = gl.Net('mb_2d')   # connected to VOUT in the full OTA

# 4 PMOS interdigitized — all gates share MA_G
p_ma1 = gl.pmos(w=6.0, fingers=2, g=ma_g, d=ma_1d, s=gl.vdd)
p_ma2 = gl.pmos(w=6.0, fingers=2, g=ma_g, d=ma_2d, s=gl.vdd)
p_mb1 = gl.pmos(w=6.0, fingers=2, g=ma_g, d=mb_1d, s=gl.vdd)
p_mb2 = gl.pmos(w=6.0, fingers=2, g=ma_g, d=mb_2d, s=gl.vdd)

chip = gl.build(p_ma1, p_ma2, p_mb1, p_mb2, name='p_block')
chip.show()
chip.drc()
chip.sim()
```
